# scMamba H5MU 数据检查
本 Notebook 只读取结构并验证配对，不会把大型稀疏矩阵转为稠密矩阵。默认路径指向本地、被 Git 忽略的 PBMC 文件。

```python
from pathlib import Path
import mudata as md
path = Path('../datasets/multiome_pbmc/raw/PBMC.h5mu')
print('存在:', path.exists(), path)
```

```python
if path.exists():
    mdata = md.read_h5mu(path, backed='r')
    print('modalities:', list(mdata.mod))
    for name, adata in mdata.mod.items():
        print(name, adata.shape, type(adata.X).__name__, adata.obs.columns.tolist())
else:
    print('请先运行 scripts/download_processed_data.py --dataset multiome_pbmc --download')
```

```python
if path.exists() and {'rna', 'atac'} <= set(mdata.mod):
    rna_ids = mdata.mod['rna'].obs_names
    atac_ids = mdata.mod['atac'].obs_names
    print('RNA/ATAC 完全同序:', rna_ids.equals(atac_ids))
```
