# PertKGE 知识图谱样本预览

论文: https://doi.org/10.1101/2024.04.08.588632

归类: 药物靶点预测／扰动转录组知识图谱

展示三元组、实体/关系映射、训练测试划分的最小预览。

```python
import sys
from pathlib import Path
import pandas as pd

base = Path('..')
print(f'项目目录: {base.resolve()}')

# 加载资源清单
resources = pd.read_csv(base / 'data_manifest' / 'resources.csv')
print(f'\n=== 资源清单 ({len(resources)} 条) ===')
print(resources[['resource_id','category','name','download_status']].to_string(index=False))
```

## 1. 三元组文件预览

```python
data_dir = base / 'data' / 'raw' / 'pertkge_package'
txt_files = list(data_dir.glob('*.txt')) if data_dir.exists() else []

if txt_files:
    print(f'找到 {len(txt_files)} 个txt文件:')
    for f in txt_files:
        print(f'  {f.name} ({f.stat().st_size/1024:.1f} KB)')
    
    # 预览 cause.txt
    cause = data_dir / 'cause.txt'
    if cause.exists():
        df = pd.read_csv(cause, sep='\t', header=None, nrows=10)
        print(f'\ncause.txt 前10行:')
        print(df.to_string(index=False, header=False))
        print(f'\n列数: {df.shape[1]}')
else:
    print('未找到数据包。请先下载 PertKGE 数据。')
    print('URL: https://drive.google.com/file/d/1jFo0dDAnUOzMoKHFqPRM4pd_loTFwmMa/view')
    print()
    print('=== 示意数据（非真实数据）===')
    print('head\trelation\ttail')
    print('DrugA\tcause\tGeneX')
    print('DrugB\tprocess\tPathwayY')
    print('GeneX\teffect\tDiseaseZ')
```

## 2. 实体/关系映射预览

```python
if txt_files:
    entity_map = data_dir / 'entity2id.txt'
    relation_map = data_dir / 'relation2id.txt'
    if entity_map.exists():
        df = pd.read_csv(entity_map, sep='\t', header=None, nrows=10)
        print(f'实体映射 (前10):')
        print(df.to_string(index=False, header=False))
    if relation_map.exists():
        df = pd.read_csv(relation_map, sep='\t', header=None, nrows=10)
        print(f'\n关系映射 (前10):')
        print(df.to_string(index=False, header=False))
else:
    print('数据包未下载，无法预览映射文件。')
```

## 3. 训练/测试划分预览

```python
if txt_files:
    test = data_dir / 'test.txt'
    if test.exists():
        df = pd.read_csv(test, sep='\t', header=None)
        print(f'测试集: {len(df)} 三元组')
        print(f'前5行:')
        print(df.head().to_string(index=False, header=False))
    else:
        print('未找到 test.txt')
else:
    print('数据包未下载。')
    print()
    print('注意: 需实际核验三元组列顺序、实体编号和关系编号映射。')
    print('不根据名称猜测文件内容。')
```
