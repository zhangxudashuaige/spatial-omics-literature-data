# 02 - 检查 Norman 2019 数据

加载 NormanWeissman2019_filtered.h5ad，检查结构、单/组合扰动分布。

```python
import sys
sys.path.insert(0, '..')
import anndata as ad
import numpy as np
import pandas as pd
from pathlib import Path

data_path = Path('../data/raw/NormanWeissman2019_filtered.h5ad')
if data_path.exists():
    adata = ad.read_h5ad(data_path)
    print(f'Shape: {adata.shape}')
    print(f'obs: {list(adata.obs.columns)}')
else:
    print(f'Data not found: {data_path} (~698.7 MB)')
    print('Run: python scripts/download_dataset.py --file NormanWeissman2019_filtered.h5ad')
```

```python
# 单扰动 vs 组合扰动统计
if 'adata' in locals():
    perturb_col = None
    for col in ['perturbation', 'condition', 'guide']:
        if col in adata.obs.columns:
            perturb_col = col
            break
    
    if perturb_col:
        counts = adata.obs[perturb_col].value_counts()
        single = sum(c for k, c in counts.items() if '+' not in str(k) and str(k).lower() not in ['ctrl','control'])
        combo = sum(c for k, c in counts.items() if '+' in str(k))
        ctrl = sum(c for k, c in counts.items() if str(k).lower() in ['ctrl','control'])
        print(f'Control: {ctrl}')
        print(f'Single perturbation: {single}')
        print(f'Combinatorial perturbation: {combo}')
        print(f'Total conditions: {len(counts)}')
```
