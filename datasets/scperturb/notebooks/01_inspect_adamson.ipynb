# 01 - 检查 Adamson 2016 数据

加载 AdamsonWeissman2016_GSM2406675_10X001.h5ad，检查结构和扰动分布。

```python
import sys
sys.path.insert(0, '..')
import anndata as ad
import numpy as np
import pandas as pd
from pathlib import Path

# 加载 Adamson 数据
data_path = Path('../data/raw/AdamsonWeissman2016_GSM2406675_10X001.h5ad')
if data_path.exists():
    adata = ad.read_h5ad(data_path)
    print(f'Shape: {adata.shape}')
    print(f'obs: {list(adata.obs.columns)}')
    print(f'var: {list(adata.var.columns)}')
else:
    print(f'Data not found: {data_path}')
    print('Run: python scripts/download_dataset.py --file AdamsonWeissman2016_GSM2406675_10X001.h5ad')
```

```python
# 扰动条件统计
if 'adata' in locals():
    perturb_col = None
    for col in ['perturbation', 'condition', 'guide', 'target']:
        if col in adata.obs.columns:
            perturb_col = col
            break
    
    if perturb_col:
        counts = adata.obs[perturb_col].value_counts()
        print(f'Perturbation column: {perturb_col}')
        print(f'Total perturbations: {len(counts)}')
        print(f'Top 20:')
        print(counts.head(20))
```

```python
# 细胞质量统计
if 'adata' in locals():
    from scipy.sparse import issparse
    if issparse(adata.X):
        umi = np.asarray(adata.X.sum(axis=1)).flatten()
        n_genes = np.asarray((adata.X > 0).sum(axis=1)).flatten()
    else:
        umi = np.asarray(adata.X.sum(axis=1)).flatten()
        n_genes = np.asarray((adata.X > 0).sum(axis=1)).flatten()
    print(f'UMI: mean={umi.mean():.1f}, median={np.median(umi):.1f}')
    print(f'Genes detected: mean={n_genes.mean():.1f}, median={np.median(n_genes):.1f}')
```
