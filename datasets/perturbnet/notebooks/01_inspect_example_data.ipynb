# 01 - 检查 PerturbNet 示例数据

加载 Hugging Face example_data 中的小型教程数据，检查结构和内容。

```python
import sys
sys.path.insert(0, '..')
from pathlib import Path
import anndata as ad
import numpy as np

# 查找示例数据
example_dir = Path('../data/example')
h5ad_files = list(example_dir.glob('**/*.h5ad'))
print(f'Example h5ad files: {[f.name for f in h5ad_files]}')

if not h5ad_files:
    print()
    print('No example data found.')
    print('Run: python scripts/inspect_huggingface.py')
    print('Then: python scripts/download_huggingface.py --phase 1')
```

```python
# 加载并检查第一个示例文件
h5ad_files = list(Path('../data/example').glob('**/*.h5ad'))
if h5ad_files:
    adata = ad.read_h5ad(h5ad_files[0])
    print(f'Loaded: {h5ad_files[0].name}')
    print(f'Shape: {adata.shape}')
    print(f'obs: {list(adata.obs.columns)}')
    print(f'var: {list(adata.var.columns)}')
    print(f'obsm: {list(adata.obsm.keys())}')
    print()
    
    # 扰动统计
    for col in ['perturbation', 'condition', 'perturb']:
        if col in adata.obs.columns:
            print(f'{col} distribution:')
            print(adata.obs[col].value_counts().head(20))
            break
```

```python
# 表达矩阵基本统计
if 'adata' in locals():
    from scipy.sparse import issparse
    X = adata.X
    print(f'X dtype: {X.dtype}')
    print(f'X sparse: {issparse(X)}')
    
    if issparse(X):
        data = X.data
        cell_totals = np.asarray(X.sum(axis=1)).flatten()
        cell_genes = np.asarray((X > 0).sum(axis=1)).flatten()
    else:
        data = X.flatten()
        cell_totals = X.sum(axis=1)
        cell_genes = (X > 0).sum(axis=1)
    
    print(f'Expression range: [{data.min():.2f}, {data.max():.2f}]')
    print(f'Cell total counts: mean={cell_totals.mean():.1f}')
    print(f'Cell detected genes: mean={cell_genes.mean():.1f}')
```
