# CellCap 人单核细胞数据分析

加载 CellCap_1MPBMC_Mono.h5ad，训练 CellCap 并导出响应程序。

```python
import anndata as ad
import scanpy as sc
import numpy as np
from pathlib import Path

data_path = Path('../data/raw/CellCap_1MPBMC_Mono.h5ad')
if data_path.exists():
    adata = ad.read_h5ad(data_path)
    print(f'Loaded: {adata.shape}')
else:
    print(f'Data not found: {data_path}')
    print('This file may require Git LFS or a separate download link.')
    print('Check the official CellCap repository for instructions.')
```

```python
# 基本 QC 和可视化
if 'adata' in locals():
    sc.pp.calculate_qc_metrics(adata, inplace=True)
    print(f'Mean counts per cell: {adata.obs["total_counts"].mean():.1f}')
    print(f'Mean genes per cell: {adata.obs["n_genes_by_counts"].mean():.1f}')
    
    # 如果有 UMAP
    if 'X_umap' in adata.obsm:
        sc.pl.umap(adata, color=['cell_type'] if 'cell_type' in adata.obs else None)
```
