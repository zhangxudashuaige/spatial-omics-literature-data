# 02 - 训练 GEARS (Norman)

使用 Norman 2019 数据训练 GEARS 模型，完成最小复现。

```python
import sys
sys.path.insert(0, '..')
from gears import PertData, GEARS

# 加载数据
pert_data = PertData('../data')
pert_data.load(data_name='norman')
print(f'Data loaded: {pert_data.adata.shape}')
```

```python
# 准备数据划分和 DataLoader
pert_data.prepare_split(split='simulation', seed=1)
pert_data.get_dataloader(batch_size=32, test_batch_size=128)
print('DataLoader ready')
print(f'Train perturbations: {len(pert_data.set2genes["train"])}')
print(f'Test perturbations: {len(pert_data.set2genes["test"])}')
```

```python
# 初始化并训练模型
import torch
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'Device: {device}')

model = GEARS(pert_data, device=device)
model.model_initialize(hidden_size=64)
print('Model initialized')

# 训练 (epochs=20 for minimal reproduction)
model.train(epochs=20)
print('Training complete')
```

```python
# 保存模型
model_path = '../models/gears_norman_hidden64_epoch20.pt'
import os
os.makedirs('../models', exist_ok=True)
torch.save(model.model.state_dict(), model_path)
print(f'Model saved: {model_path}')
```
