# 01 - 检查 Norman 2019 数据

加载 GEARS 官方预处理的 Norman 数据，检查结构、扰动条件和细胞分布。

```python
import sys
sys.path.insert(0, '..')
from gears import PertData
import anndata as ad
import pandas as pd
import numpy as np

# 加载 Norman 数据
pert_data = PertData('../data')
pert_data.load(data_name='norman')
adata = pert_data.adata
print(f'Shape: {adata.shape}')
print(f'obs: {list(adata.obs.columns)}')
print(f'var: {list(adata.var.columns)}')
```

```python
# 扰动条件统计
conditions = adata.obs['condition'].value_counts()
print(f'Total conditions: {len(conditions)}')
print(f'Control cells: {conditions.get("ctrl", 0)}')

single = sum(c for k, c in conditions.items() if '+' not in k and k != 'ctrl')
combo = sum(c for k, c in conditions.items() if '+' in k)
print(f'Single perturbation cells: {single}')
print(f'Combinatorial perturbation cells: {combo}')
print()
print('Top 20 conditions:')
print(conditions.head(20))
```

```python
# 基因和表达矩阵检查
print(f'Genes: {adata.n_vars}')
print(f'Gene names complete: {adata.var_names.is_unique and not adata.var_names.str.empty.any()}')
print(f'X dtype: {adata.X.dtype}')
from scipy.sparse import issparse
print(f'X sparse: {issparse(adata.X)}')

# 细胞质量
if issparse(adata.X):
    totals = np.asarray(adata.X.sum(axis=1)).flatten()
    n_genes = np.asarray((adata.X > 0).sum(axis=1)).flatten()
else:
    totals = np.asarray(adata.X.sum(axis=1)).flatten()
    n_genes = np.asarray((adata.X > 0).sum(axis=1)).flatten()
print(f'Mean total counts: {totals.mean():.1f}')
print(f'Mean detected genes: {n_genes.mean():.1f}')
```
