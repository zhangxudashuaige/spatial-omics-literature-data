# 02 检查表达数据
自动寻找 `data/processed/` 中的 H5AD、GEF 或 GEM。没有真实文件时只说明缺口，不伪造结构。

```python
from pathlib import Path
import json
import subprocess
import sys
project = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
patterns = ('*.h5ad', '*.gef', '*.gem', '*.gem.gz', '*.csv', '*.tsv')
files = [p for pattern in patterns for p in (project / 'data/processed').rglob(pattern)]
files[:10]
```

```python
if files:
    target = min(files, key=lambda p: p.stat().st_size)
    report = project / 'results/tables/expression_schema.json'
    subprocess.run([sys.executable, str(project/'scripts/inspect_spatial_files.py'), str(target), '--json', str(report)], check=True)
    schema = json.loads(report.read_text(encoding='utf-8'))
else:
    schema = {'status': '没有本地处理后表达文件', 'next_step': '按 docs/download_guide.md 放入作者公开的处理后文件后重跑'}
schema
```
