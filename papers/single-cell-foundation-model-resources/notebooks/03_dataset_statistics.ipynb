# 03 下游数据统计
查看任务类别、物种、访问方式和原始论文覆盖情况。

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
ds = pd.read_csv(ROOT / 'metadata' / 'downstream_datasets.csv').fillna('')
summary = {
    '数据集数': len(ds),
    '有原始论文链接': int(ds.original_paper_url.ne('').sum()),
    '有accession': int(ds.accession.ne('').sum()),
    '处理后数据可用': int(ds.processed_data_available.eq('yes').sum()),
    '原始数据可用': int(ds.raw_data_available.eq('yes').sum()),
}
pd.Series(summary, name='count')
```

```python
cats = ds.dataset_category.str.split('; ').explode().value_counts().sort_values()
ax = cats.plot.barh(figsize=(8, 4), color='#59A14F')
ax.set(xlabel='唯一数据集数', ylabel='类别', title='下游数据类别')
plt.tight_layout()
```
