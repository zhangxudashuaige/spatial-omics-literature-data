# 02 模型—数据—任务关系
一个数据集只在数据表登记一次；多模型、多任务关系由矩阵表表达。

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
rel = pd.read_csv(ROOT / 'metadata' / 'task_model_matrix.csv')
counts = rel.groupby('model_name').size().sort_values()
ax = counts.plot.barh(figsize=(8, 7), color='#4C78A8')
ax.set(xlabel='登记关系数', ylabel='模型', title='各模型关联的语料/下游数据数量')
plt.tight_layout()
counts.tail(10)
```
