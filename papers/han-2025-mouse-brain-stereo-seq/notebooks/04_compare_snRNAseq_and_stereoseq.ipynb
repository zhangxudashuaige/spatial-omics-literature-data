# 04 比较 snRNA-seq 与 Stereo-seq

把 Stereo-seq 按 cluster 聚合后，与合成 snRNA-seq cluster 参考表达比较。

> 本 Notebook 默认使用合成测试数据，不代表论文真实数值。

```python
from pathlib import Path
ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
SAMPLE = ROOT / 'data' / 'sample'
RESULTS = ROOT / 'results'
print('项目目录:', ROOT)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
expression = pd.read_csv(SAMPLE / 'sample_spatial_expression.csv')
metadata = pd.read_csv(SAMPLE / 'sample_cell_metadata.csv')[['cell_id','cell_cluster']]
reference = pd.read_csv(SAMPLE / 'sample_snrna_reference.csv')
spatial = expression.merge(metadata, on='cell_id').groupby(['cell_cluster','gene'], as_index=False)['umi_count'].mean()
comparison = spatial.merge(reference, on=['cell_cluster','gene'], how='inner')
comparison
```

```python
fig, ax = plt.subplots(figsize=(6,5))
for gene, group in comparison.groupby('gene'):
    ax.scatter(group.mean_expression, group.umi_count, label=gene, s=60)
ax.set(xlabel='snRNA-seq mean expression (synthetic)', ylabel='Stereo-seq mean UMI (synthetic)', title='Cluster-level comparison')
ax.legend(); fig.tight_layout(); plt.show()
```
