# 03 可视化二维空间基因表达

按 cell_id 合并 Gad1 表达与二维坐标，并保存空间散点图。

> 本 Notebook 默认使用合成测试数据，不代表论文真实数值。

```python
from pathlib import Path
ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
SAMPLE = ROOT / 'data' / 'sample'
RESULTS = ROOT / 'results'
print('项目目录:', ROOT)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
expression = pd.read_csv(SAMPLE / 'sample_spatial_expression.csv')
metadata = pd.read_csv(SAMPLE / 'sample_cell_metadata.csv')
gad1 = expression.query("gene == 'Gad1'")[["cell_id","umi_count"]]
plot_data = metadata.merge(gad1, on='cell_id', validate='one_to_one')
plot_data.head()
```

```python
out = RESULTS / 'figures' / 'notebook_Gad1_spatial_expression.png'
out.parent.mkdir(parents=True, exist_ok=True)
fig, ax = plt.subplots(figsize=(7,5))
sc = ax.scatter(plot_data.x, plot_data.y, c=plot_data.umi_count, s=90, cmap='viridis', edgecolor='white')
ax.set(title='Synthetic sample: Gad1 spatial expression', xlabel='slice x', ylabel='slice y')
ax.invert_yaxis(); ax.set_aspect('equal', adjustable='box')
fig.colorbar(sc, ax=ax, label='UMI count')
fig.tight_layout(); fig.savefig(out, dpi=180); plt.show()
print(out)
```
