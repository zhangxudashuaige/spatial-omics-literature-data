# 05 可视化 CCF 三维坐标

用合成 CCF 坐标展示三维散点；真实 CCF 文件取得后可替换输入。

> 本 Notebook 默认使用合成测试数据，不代表论文真实数值。

```python
from pathlib import Path
ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
SAMPLE = ROOT / 'data' / 'sample'
RESULTS = ROOT / 'results'
print('项目目录:', ROOT)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
metadata = pd.read_csv(SAMPLE / 'sample_cell_metadata.csv')
```

```python
fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')
for region, group in metadata.groupby('brain_region'):
    ax.scatter(group.ccf_x, group.ccf_y, group.ccf_z, label=region, s=45)
ax.set(xlabel='CCF x', ylabel='CCF y', zlabel='CCF z', title='Synthetic CCF coordinates')
ax.legend(bbox_to_anchor=(1.04,1), loc='upper left', fontsize=8)
fig.tight_layout(); plt.show()
```
