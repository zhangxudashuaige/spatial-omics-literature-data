# 检查SpaGCN输入
读取本地官方151673表达矩阵和坐标，并检查barcode对齐；不会读取大图或稠密化完整矩阵。

```python
from pathlib import Path
root = Path('../datasets/official_toy_data/raw/151673')
h5 = root / 'expression_matrix.h5'
coords = root / 'positions.txt'
print(h5.exists(), coords.exists())
```

```python
import h5py, pandas as pd
if h5.exists():
    with h5py.File(h5) as f:
        print('H5 gene×spot:', f['matrix/shape'][:].tolist())
        barcodes = {x.decode() for x in f['matrix/barcodes'][:]}
    pos = pd.read_csv(coords, header=None)
    print('坐标行数:', len(pos), '完全匹配:', barcodes == set(pos[0]))
else:
    print('请按README下载官方151673输入')
```
